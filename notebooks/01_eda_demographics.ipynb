# 01 — Exploration démographique de la cohorte

Objectif : comprendre la composition de la population simulée (âge, sexe, répartition géographique) avant l'analyse clinique.

```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt

from etl import build_patient_dataset

df = build_patient_dataset('../data/synthea_output')
df.head()
```

```python
print(f"Nombre de patients : {len(df)}")
df['gender'].value_counts()
```

```python
df['age'].plot(kind='hist', bins=20, title='Distribution des âges')
plt.xlabel('Âge')
plt.show()
```

```python
df['city'].value_counts().head(10).plot(kind='barh', title='Top 10 villes')
plt.show()
```

## Observations

Sur les **150 patients** analysés :

- La population est légèrement plus masculine (80 hommes / 70 femmes), avec un âge moyen de **50,1 ans** (de 19 à 80 ans) — une pyramide des âges relativement équilibrée mais avec une part significative de patients de plus de 60 ans, tranche particulièrement pertinente pour le suivi des maladies chroniques.
- La population est répartie sur plusieurs villes, avec une concentration plus forte à Toamasina (30 patients), Antsirabe (28) et Toliara (25) — utile pour cibler d'éventuelles campagnes de dépistage localisées.
- Cette structure démographique équilibrée en âge permet une analyse fiable de la relation entre âge et risque clinique dans les notebooks suivants.
